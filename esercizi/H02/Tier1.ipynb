# Calcolo di $\pi$

```python
import random

N = 1000000 # Numero di lanci
i = 0 # Dentro il cerchio
o = 0 # Fuori dal cerchio


for n in range(N): # Genera e controlla se sono dentro a cerchio
    x = random.uniform(-1, 1)
    y = random.uniform(-1, 1)
    r = x**2 + y**2 
    if r <= 1:
        i += 1
    else:
        o += 1

print(f"Dentro il cerchio: {i:d}")
print(f"Fuori dal cerchio: {o:d}")

pi = 4*i/N # Stima di π

print("Valore di π: ", pi)
```

# Calcolo di $\pi$ con media su più simulazioni

```python
import random
from math import sqrt
random.seed(67)

L = [] # Stime per ogni simulazione
K = 20 # Numero di simulazioni
N = 10000 # Numero di lanci


for m in range(K): # Avvia le K simulazioni

    i = 0 # Dentro il cerchio
    o = 0 # Fuori dal cerchio

    for n in range(N): # Genera e controlla se sono dentro a cerchio
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)
        r = x**2 + y**2 
        if r <= 1:
            i += 1
        else:
            o += 1
        
    print(f"-----------Simulazione {m+1}-----------")
    print(f"Dentro il cerchio: {i}")
    print(f"Fuori dal cerchio: {o}")
    pi = 4*i/N # Stima di π
    print("Valore di π: ", pi)

    L.append(pi) # Aggiungo il valore della simulazione
    
print(L)

# Calcolo di media, deviazione standard e sdom
media = sum(L)/K
somma_scarti = sum((z - media)**2 for z in L)
std = sqrt(1/(K-1)*somma_scarti)
sdom = std/sqrt(K)

print(f"Media: {media:.5f}")
print(f"Deviazione standard: {std:.5f}")
print(f"SDOM: {sdom:.5f}")
```

Aumentando N, cioè il numero di iterazioni, la stima di pi greco si stabilizza rendendo le K stime più vicine tra loro. Questo ha l'effetto di dimuniera la deviazione standard e quindi l'errore statistico.
